```python
#####Task 12 : Store inventory


products = {}

def add_product():
    name = input("Product Name: ")
    price = float(input("Price: "))
    quantity = int(input("Quantity: "))

    products[name] = {
        "price": price,
        "quantity": quantity
    }

    print("Product added successfully.")


def sell_product():
    name = input("Product Name: ")
    quantity = int(input("Quantity to sell: "))

    if name in products:
        if products[name]["quantity"] >= quantity:
            products[name]["quantity"] -= quantity
            print("Product sold successfully.")
        else:
            print("WARNING: Insufficient stock.")
    else:
        print("Product not found.")


def restock_product():
    name = input("Product Name: ")
    quantity = int(input("Quantity to add: "))

    if name in products:
        products[name]["quantity"] += quantity
        print("Product restocked successfully.")
    else:
        print("Product not found.")


def search_product():
    name = input("Product Name: ")

    if name in products:
        print("Product:", name)
        print("Price:", products[name]["price"])
        print("Quantity:", products[name]["quantity"])
    else:
        print("Product not found.")


def display_inventory():
    print("\n--- Inventory ---")

    total_value = 0

    for name, product in products.items():
        price = product["price"]
        quantity = product["quantity"]
        value = price * quantity

        print(name, "| Price:", price, "| Quantity:", quantity, "| Value:", value)

        total_value += value

    print("Total Inventory Value:", total_value)

    print("\n--- Low Stock Products ---")

    for name, product in products.items():
        if product["quantity"] < 5:
            print(name, "| Quantity:", product["quantity"])


while True:
    print("\n1. Add Product")
    print("2. Sell Product")
    print("3. Restock Product")
    print("4. Search Product")
    print("5. Display Inventory")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_product()
    elif choice == "2":
        sell_product()
    elif choice == "3":
        restock_product()
    elif choice == "4":
        search_product()
    elif choice == "5":
        display_inventory()
    elif choice == "6":
        print("Program ended.")
        break
    else:
        print("Invalid choice.")
```

```python
# Master list patient records (dictionaries) save karne ke liye
appointments = []

def get_patient_by_id(p_id):
    """ID ke zariye patient search karne ka helper function"""
    for patient in appointments:
        if patient["id"] == p_id:
            return patient
    return None

def add_appointment():
    print("\n--- Naya Appointment Add Karein ---")
    p_id = input("Patient ID enter karein: ").strip()

    # Requirement 4: Duplicate ID check
    if get_patient_by_id(p_id):
        print("Error: Ye Patient ID pehle se majood hai! Unique ID use karein.")
        return

    name = input("Patient ka Naam: ").strip()
    age = input("Umar (Age): ").strip()
    disease = input("Bimari / Masla (Disease): ").strip()
    time = input("Appointment Time (e.g., 10:30 AM): ").strip()

    is_emergency = input("Kya ye Emergency hai? (yes/no): ").strip().lower()
    emergency_status = True if is_emergency in ['yes', 'y'] else False

    # Requirement 2: Patient info ko dictionary mein store karna
    patient = {
        "id": p_id,
        "name": name,
        "age": age,
        "disease": disease,
        "time": time,
        "emergency": emergency_status
    }

    # Requirement 3: Main list mein save karna
    appointments.append(patient)
    print("Appointment kamyabi se add ho gaya hai!")

def search_patient():
    print("\n--- Patient Record Search Karein ---")
    p_id = input("Search karne ke liye Patient ID enter karein: ").strip()
    patient = get_patient_by_id(p_id)

    # Requirement 5: Patient ID se search
    if patient:
        status = "EMERGENCY"
        if patient["emergency"]
        else "Normal"
        print(f"\n[ID: {patient['id']}] Naam: {patient['name']} | Umar: {patient['age']}")
        print(f"Masla: {patient['disease']} | Time: {patient['time']} | Status: {status}")
    else:
        print("Is ID ka koi record nahi mila.")

def display_appointments():
    print("\n--- Tamam Appointments ---")
    if not appointments:
        print("Koi appointment majood nahi hai.")
        return

    # Requirement 6 & 7: Emergency patients ko pehle filter karke top par dikhana
    emergency_list = [p for p in appointments if p["emergency"]]
    normal_list = [p for p in appointments if not p["emergency"]]

    sorted_appointments = emergency_list + normal_list

    for index, p in enumerate(sorted_appointments, 1):
        status = "EMERGENCY" if p["emergency"] else "Normal"
        print(f"{index}. [ID: {p['id']}] {p['name']} (Umar: {p['age']}) - {p['disease']} | Time: {p['time']} | [{status}]")

def cancel_appointment():
    print("\n--- Appointment Cancel Karein ---")
    p_id = input("Cancel karne ke liye Patient ID enter karein: ").strip()
    patient = get_patient_by_id(p_id)

    # Requirement 8: Appointment cancel karna
    if patient:
        appointments.remove(patient)
        print(f"Patient ID {p_id} ({patient['name']}) ka appointment cancel ho gaya hai.")
    else:
        print("Patient ID nahi mili.")

def display_summary():
    print("\n--- Clinic Summary ---")
    total_patients = len(appointments)

    # Requirement 9: Emergency aur Normal patients count karna
    emergency_count = sum(1 for p in appointments if p["emergency"])
    normal_count = total_patients - emergency_count

    print(f"Kul Registered Patients: {total_patients}")
    print(f"Emergency Patients: {emergency_count}")
    print(f"Normal Patients: {normal_count}")

def main_menu():
    """Requirement 10: Menu driven program using simple logic"""
    while True:
        print("\n==============================")
        print("  Clinic Appointment Manager")
        print("==============================")
        print("1. Add Appointment")
        print("2. Search Patient")
        print("3. Display Appointments")
        print("4. Cancel Appointment")
        print("5. Display Clinic Summary")
        print("6. Exit")

        choice = input("Option select (1-6): ").strip()

        if choice == '1':
            add_appointment()
        elif choice == '2':
            search_patient()
        elif choice == '3':
            display_appointments()
        elif choice == '4':
            cancel_appointment()
        elif choice == '5':
            display_summary()
        elif choice == '6':
            print("Program band ho raha hai. Allah Hafiz!")
            break
        else:
            print("Wrong option! Select the number from 1 to 6")

main_menu()
```
