```python
import pandas as pd
#task 1
data = {
    "name": ['Ali' , 'Sara', 'Ahmed' ],
     "Age": [20, 21, 19],
    "marks":[80, 90, 75]
    }
df = pd.DataFrame(data)

#Task 2
print("Average marks:", df['marks'].mean())
print("Max marks:", df['marks'].max())
print("Min marks:", df['marks'].min())
print("Total students:", len(df))
print("Students greater than or equal to 50:\n",df[df["marks"] >= 50])
print("Students less than or equal to 50:\n",df[df["marks"] <= 50])

#task 3
df["Cities"] = ["Karachi", "Lahore", "Islamabad"]

print(df)
```
